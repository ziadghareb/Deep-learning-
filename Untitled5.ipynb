```python
import numpy as np

# Tokenized sentence
sentence = ["What", "are", "the", "symptoms", "of", "diabetes"]

# Step 1: Simple fixed embeddings (instead of random)
embeddings = np.array([
    [1, 0, 1],
    [0, 1, 1],
    [1, 1, 0],
    [0, 0, 1],
    [1, 0, 0],
    [0, 1, 0]
])

# Step 2: Weight matrices (simplified)
Wq = np.array([[1, 0, 1],
               [0, 1, 0],
               [1, 0, 1]])

Wk = np.array([[0, 1, 0],
               [1, 0, 1],
               [0, 1, 0]])

Wv = np.array([[1, 1, 0],
               [0, 1, 1],
               [1, 0, 1]])

# Step 3: Compute Q, K, V
Q = np.dot(embeddings, Wq)
K = np.dot(embeddings, Wk)
V = np.dot(embeddings, Wv)

# Step 4: Attention scores
scores = np.dot(Q, K.T)

# Step 5: Normalize scores (softmax)
def softmax(x):
    e_x = np.exp(x - np.max(x))
    return e_x / e_x.sum(axis=1, keepdims=True)

attention = softmax(scores)

# Step 6: Final output
output = np.dot(attention, V)

print("Attention Scores:\n", attention)
print("\nFinal Output:\n", output)
```
